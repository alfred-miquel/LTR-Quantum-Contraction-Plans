```julia
"""
    parse_contraction_plan_to_tuples(filename::String)
Read a contraction plane file and convert it to the same format as 
usual in minfill  or FlowCutter, returning a Vector{Tuple{Symbol, Symbol, Symbol}}.

Format of each line:
    t1 t2 t3 r1 r2 r3 indices

Where:
    - t1, t2, t3: names of tensors (first 3 words) that become Symbol
    - The rest of the information is ignored for this format

# Return:
    - Vector{Tuple{Symbol, Symbol, Symbol}}: Every tuple is (tensor1, tensor2, result)
"""



function parse_contraction_plan_to_tuples(filename::String)
    contraction_steps = Vector{Tuple{Symbol, Symbol, Symbol}}()
    step_count = 0
    
    open(filename, "r") do file
        for line in eachline(file)
            # Jump empty lines
            isempty(strip(line)) && continue
            
            # split the line by spaces
            parts = split(line)
            
            # three words are required
            if length(parts) < 3
                @warn "Line with incorrect format (few elements): $line"
                continue
            end
            
            # Convert to a symbol the name of tensors
            tensor1 = Symbol(parts[1])
            tensor2 = Symbol(parts[2])
            result_tensor = Symbol(parts[3])
            
            # Add contraction to the list
            push!(contraction_steps, (tensor1, tensor2, result_tensor))
            step_count += 1
        end
    end
    
    println("Processed $step_count steps of contraction")
    
    return contraction_steps
end
```

```julia
begin
     

    directori = "../Data/plans"
    
    nom_v = []

    plans_v =[]


    circuit_v = ["resultats_QFT_5_pla_fc.txt","resultats_QFT_5_pla_gn.txt","resultats_QFT_5_pla_jdrasil.txt","resultats_QFT_5_pla_mf.txt","resultats_QFT_5_pla_minfill_mrs.txt","resultats_QFT_5_pla_minfillbg_mrs.txt","resultats_QFT_5_pla_tamaki.txt"]        
  
    for circuit in circuit_v
    
          try
            filename= directori* "/" * circuit # if you are using Windows, you should exchange "/" by "\\"
            plan = parse_contraction_plan_to_tuples(filename)
            push!(plans_v, plan)

           catch
               println("Error in $circuit")
           end
        
      
    
        end

end
```
